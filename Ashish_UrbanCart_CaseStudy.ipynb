```python
#1
def get_product_price(product_id,products_list):
  for i in products_list:
    if i["Product_ID"]==product_id:
      return i["Price"]

  return None;
```

```python
#2
def get_customer_name(customer_id,customers_list):
  for i in customers:
    if i["Customer_ID"]==customer_id:
      return i["Name"]
  return None
```

```python
#3
def calculate_order_value(price,quantity):
  return  price*quantity
```

```python
#4
def classify_order(value,high_limit=3000,low_limit=1000):
  if value >=high_limit:
    return "High"
  elif value >=low_limit and value < high_limit:
    return "Medium"
  else:
    return "Low"
  return None
```

```python
#5
def generate_invoice(order, products_list, customers_list):
  price=get_product_price(order["Product_ID"], products_list)
  name=get_customer_name(order["Customer_ID"], customers_list)

  product_name=None
  for product in products_list:
    if product["Product_ID"] == order["Product_ID"]:
      product_name = product["Name"]
      break

  order_value=calculate_order_value(price, order["Quantity"])

  invoice=f"""
    ==============================
    Order ID      : {order['Order_ID']}
    Customer      : {name}
    Product       : {product_name}
    Quantity      : {order['Quantity']}
    Order Value   : ₹{order_value}
    ==============================
    """
```

```python
#6
def calculate_total_revenue(orders_list, products_list):
  revenue=0
  for order in orders_list:
    if order["Status"]=="Cancelled":
      continue
    price=get_product_price(order["Product_ID"],products_list)
    revenue+=calculate_order_value(price,order["Quantity"])

  return revenue


```

```python
#7
def top_customer_by_spend(orders_list, products_list, customers_list):
  customer={}
  for order in orders_list:
    if order["Status"]=="Delivered":
      continue

    price=get_product_price(order["Product_ID"],products_list)
    value=calculate_order_value(price,order["Quantity"])

    customer_id=order["Customer_ID"]

    if customer_id not in customer:
      customer[customer_id]=0
    customer[customer_id]+=value

    top_customer=0
    highest_spend=0

    for customer_id,spending in customer.items():
      if spending> highest_spend:
        top_customer=customer_id
        highest_spend=spending

    return get_customer_name(top_customer, customers_list)





```

```python
#8
def apply_offer(price, *discounts):
  final_price=price

  for i in discounts:
    final_price=final_price-(final_price * i /100)

  return final_price

```
